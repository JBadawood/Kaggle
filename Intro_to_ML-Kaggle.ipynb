## **1- How Models Work**
The first step if you're new to machine learning.\
https://www.kaggle.com/code/dansbecker/how-models-work

قبل بناء أي نموذج نحتاج بيانات سابقة تم جمعها. مثلًا:

لو نبغى نتوقع أسعار البيوت الجديدة راح يتطلب مننا قبلها أننا نعرف كم كانت أسعار البيوت القديمة وكذلك الموجودة حاليًا (يعني نشوف الـ history)

بهذه الطريقة راح نعرف الـ patterns لهذه الأسعار السابقة ومن ثم يمكننا الاستفادة منها لعمل predictions للبيوت الجديدة

We'll start with a model called the **Decision Tree**. There are fancier models that give more accurate predictions. But decision trees are easy to understand, and they are *the basic building block for some of the best models* in **data science**.

> You can capture more factors using a tree that has more "**splits**." The most factors affecting home price, are: 
- number of bathrooms, 
- lot size, 
- location, etc.

> These are called **"deeper" trees**. The predicted price for the house is *at the bottom of the tree*. The point at the bottom where we make a prediction is called a **leaf**.

We use data to decide how to break the houses into two groups, and then again to determine the predicted price in each group. This step of capturing patterns from data is called **fitting** or **training** the model. The data used to **fit** the model is called the **training data**.

After the model has been fit, you can apply it to new data to **predict** prices of additional homes.

### 4-1 What is Model Validation

كيف نتحقق من جودة نموذج تعلم الآلة؟

أنسب مقياس لجودة النموذج هو الدقة التنبؤية predictive accuracy

يعني المطلوب معرفة هل التوقعات حق النموذج قريبة من اللي يصير في الواقع

Many people make a **huge mistake** when measuring predictive accuracy. They make predictions with their *training data* and compare those predictions to the target values in the *training data*. You'll see the problem with this approach and how to solve it.

توجد العديد من المقاييس لتلخيص جودة النموذج و واحدة منها تسمى Mean Absolute Error (MAE)



```
error = actual - predicted
```



> بس نحن راح ناخذ القيمة المطلقة لهذا الـ error


Absolute value of error
```
absolute error = |actual - predicted|
```



ومنها راح نجمع كل القيم حق الـ absolute error وبعدها راح ناخذ المتوسط حقهم

<a name='sec4-2'></a>
### 4-2  Calculate Mean Absolute Error (MAE)

````python
import pandas as pd
````

````python
from google.colab import drive
drive.mount('/content/drive')
````

````python
melbourne_data = pd.read_csv('/content/drive/MyDrive/iMLND/Kaggle/Intro_to_ML/melb_data.csv')
````

````python
# Filter rows with missing price values
filtered_melbourne_data = melbourne_data.dropna(axis = 0)
````

````python
# Choose target and features
y = filtered_melbourne_data.Price

melbourne_features = ['Rooms', 'Bathroom', 'Landsize', 'BuildingArea',
                      'YearBuilt', 'Lattitude', 'Longtitude']
X = filtered_melbourne_data[melbourne_features]
````

````python
from sklearn.tree import DecisionTreeRegressor
````

````python
# Define model
melbourne_model = DecisionTreeRegressor()
````

````python
# Fit model
melbourne_model.fit(X,y)
````

````python
# Calculate the mean absolute error
from sklearn.metrics import mean_absolute_error

predicted_home_prices = melbourne_model.predict(X)

mean_absolute_error(y, predicted_home_prices)
````

<a name='sec4-3'></a>
### 4-3 The Problem with "In-Sample" Scores

في الخطوة السابقة كما لاحظنا أن البيانات اللي استخدمناها في تدريب النموذج هي نفسه اللي قيمنا بها النموذج وهذا شي غير مقبول

السبب هو أن النموذج راح يطلع bias وراح تطلع لنا وكأن النتائج ودقة النموذج عالية

لكن لو جربنا النموذج على بيانات جديدة راح يفشل في التنبؤ بشكل سليم

لذلك يجب عند قياس اداء النموذج أن نستخدم بيانات جديدة لم يتم استخدامها أثناء بناء النموذج

وحدة من الطرق هي أننا نقسم البيانات إلى:
- training data
- **validation data** (لاختبار دقة النموذج)

في مكتبة scikit-learn توجد function باسم `train-test-split` لتقسم البيانات اللي لدينا:
-  **training data**: to fit the model
-  **validation data**: to calculate mean_absolute_error

هذا التقسم راح يتم على كلا الـ features والـ target

````python
from sklearn.model_selection import train_test_split
````

تقسيم البيانات راح يكون بشكل عشوائي بس نحن هنا راح نقيد عملية التقسيم ليكون التقسيم دائمًا نفسه عبر استخدام `random_state` argument

````python
train_X, val_X, train_y, val_y = train_test_split(X, y, random_state = 0)
````

````python
# Define model
melbourne_model = DecisionTreeRegressor()
````

````python
# Fit model
melbourne_model.fit(train_X, train_y)
````

````python
# Get predicted prices on validation data
val_predictions = melbourne_model.predict(val_X)
````

````python
mean_absolute_error(val_y, val_predictions)
````

> هنا نلاحظ الفرق بين القيمة اللي طلعناها سابقًا لمن استخدمنا in-sample data حيث كانت الـ MAE حوالي 434

> بس لمن استخدمنا الـ out-sample data طلعت لنا الـ MAE بحوالي 278000

## **5- Underfitting and Overfitting**
Optimizing the Model 

Fine-tune your model for better performance.\
https://www.kaggle.com/code/dansbecker/underfitting-and-overfitting

عشان نحسن من دقة النموذج مثل decision tree

في هذا النوع نحتاج نتعمق ونتفرع أكثر عشان نحصل على prediction مناسبة

If a tree only had 1 split, it divides the data into 2 groups. If each group is split again, we would get 4 groups of houses. Splitting each of those again would create 8 groups. If we keep doubling the number of groups by adding more splits at each level, we'll have  $2^{10}$  groups of houses by the time we get to the 10th level. That's 1024 leaves.

زيادة التفرعات قد تقودنا للـ overfitting

يعني راح تكون جدًا دقيقة مع بيانات التدريب لكن لمن نستخدم النموذج مع بيانات جديدة راح يكون دقته جدًا ضعيفة

طيب لو قللنا التفرعات ايش ممكن يصير؟

لو كانت التفرعات قليلة جدًا هنا راح نقع في مشكلة ثانية وهي underfitting

هنا راح تكون نتائج التوقع بعيدة جدًا حتى عن بيانات التدريب وذلك بسبب أن النموذج فشل في تعلم الفروقات والأنماط المهمة من بيانات المعطى له لذلك أصبح أداءه ضعيف مع بيانات التدريب

#### Utility Function
راح تساعدنا عشان نقارن النتائج من MAE بعد تغيير الـ leaf nodes في كل مرة

````python
from sklearn.metrics import mean_absolute_error
from sklearn.tree import DecisionTreeRegressor
````

````python
def get_mae(max_leaf_nodes, train_X, val_X, train_y, val_y):
    model = DecisionTreeRegressor(max_leaf_nodes = max_leaf_nodes, random_state = 0)
    model.fit(train_X, train_y)
    preds_val = model.predict(val_X)
    mae = mean_absolute_error(val_y, preds_val)

    return(mae)
````
